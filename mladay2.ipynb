# 1. arange() - create a data with pattern 
# syntax : arange(start , stop , step)
# start : default = 0 ,stop = need to mention , step : default = 1

```python
import numpy as np
```

```python
np.arange(7)
```

```python
np.arange(1,20,2)
```

```python
np.arange(0,101)
```

```python
np.arange(2,dtype=bool)
```
